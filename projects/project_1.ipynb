# 1. NYC MTA Subway Rider Data Analysis using Pandas

## (1) Introduction on Used MTA Dataset


This mini project explores trends in New York City subway ridership using an open dataset.
The dataset contains daily ridership counts across the MTA network. 
We’ll use `pandas` to compute basic descriptive statistics like mean, median, and mode, with interpretation on what these numbers tell us about subway usage.

Dataset: New York City MTA: Number of Subway Riders (CSV)
Source: [CEIC Data](https://insights-ceicdata-com.ezproxy.cul.columbia.edu/b6f889cb-6745-4f91-a618-671f3ae47e0c/views)

## (2) Load and Read Data using Pandas

```python
import pandas as pd

# Load dataset
df = pd.read_csv("New York City MTA No of Subway Riders.csv")

# Display the first few rows
df.head()
```

## (3) Compute Mean, Median, and Mode

```python
# Select the numeric column
col = "New York City MTA: No. of Subway Riders"

mean_val = df[col].mean()
median_val = df[col].median()
mode_val = df[col].mode()[0]

print(f"Mean ridership: {mean_val:,.0f}")
print(f"Median ridership: {median_val:,.0f}")
print(f"Mode ridership: {mode_val:,.0f}")
```

## (4) Summary of Findings

- **Mean ridership** gives us the average number of subway riders per period (likely per day or week).  
- **Median ridership** shows the midpoint, reducing the effect of extreme high or low days.  
- **Mode ridership** reveals the most frequently occurring ridership count.

If the mean is higher than the median, that indicates occasional high-ridership spikes like during holidays or events. If they are close, ridership patterns are relatively consistent.


# 2. The Hard Way

```python
import csv

filename = "New York City MTA No of Subway Riders.csv"

values = []
with open(filename, newline='') as csvfile:
    reader = csv.reader(csvfile)
    header = next(reader)  
    for row in reader:
        try:
            value = int(row[1].replace(",", "")) 
            values.append(value)
        except (ValueError, IndexError):
            continue  

#Mean
total = sum(values)
count = len(values)
mean_val = total / count

#Median
sorted_values = sorted(values)
mid = count // 2
if count % 2 == 0:
    median_val = (sorted_values[mid - 1] + sorted_values[mid]) / 2
else:
    median_val = sorted_values[mid]

#Mode
counts = {}
for v in values:
    counts[v] = counts.get(v, 0) + 1
mode_val = max(counts, key=counts.get)

print(f"Number of records: {count}")
print(f"Mean ridership: {mean_val:,.0f}")
print(f"Median ridership: {median_val:,.0f}")
print(f"Mode ridership: {mode_val:,.0f}")
```

# 3. Data Visualization

```python
import csv

filename = "New York City MTA No of Subway Riders.csv"

yearly_totals = {}

with open(filename, newline='') as csvfile:
    reader = csv.reader(csvfile)
    header = next(reader)
    for row in reader:
        try:
            date = row[0]      
            value = int(row[1].replace(",", ""))
            year = date.split("/")[-1]
            yearly_totals[year] = yearly_totals.get(year, 0) + value
        except (ValueError, IndexError):
            continue

max_val = max(yearly_totals.values())

print("NYC Subway Ridership by Year\n")
for year in sorted(yearly_totals.keys()):
    count = yearly_totals[year]
    bar_length = int((count / max_val) * 40)
    bar = "█" * bar_length
    print(f"{year}: {bar} {count:,}")
```
