## Hypothesis
### Years with more fire-related emergency incidents reported to OEM will also show more confirmed fires in FDNY’s Bureau of Fire Investigations dataset, reflecting consistent trends across the two systems.

```python
import pandas as pd

oem = pd.read_csv("Emergency_Response.csv")
bfi = pd.read_csv("Bureau_of_Fire_Investigations.csv")
```

### Data cleaning for Emergency_Response

```python
fire_oem = oem[oem["Incident Type"].str.contains("FIRE", case=False, na=False)]

fire_oem["Creation Date"] = pd.to_datetime(fire_oem["Creation Date"])
fire_oem["year"] = fire_oem["Creation Date"].dt.year

oem_yearly = fire_oem.groupby("year").size().reset_index(name="oem_fire_incidents")
oem_yearly
```

### Data cleaning for Fire Inspection

```python

bfi_yearly = bfi.groupby("Case_Year").size().reset_index(name="bfi_confirmed_fires")

bfi_yearly = bfi_yearly.rename(columns={"Case_Year": "year"})
bfi_yearly
```

### Merge by year

```python
merged = pd.merge(oem_yearly, bfi_yearly, on="year")
merged
```

### Data Visualization

```python
import plotly.io as pio

pio.renderers.default = "notebook_connected+plotly_mimetype"
import plotly.express as px

fig = px.scatter(
    merged,
    x="oem_fire_incidents",
    y="bfi_confirmed_fires",
    text="year",
    labels={
        "oem_fire_incidents": "OEM Fire-Related Incidents",
        "bfi_confirmed_fires": "FDNY Confirmed Fire Investigations",
    },
    title="Relationship Between OEM Fire Incidents and FDNY Fire Investigations (2016–2024)",
    
)

fig.show()

```

## Explanation

The scatterplot shows no strong linear relationship between OEM fire-related emergency incidents and FDNY confirmed fire investigations from 2016 to 2024. Instead, the data forms two distinct clusters: a higher-activity period before COVID-19 (2017–2019) and a lower-activity period during the pandemic years (2020–2022). This suggests that both OEM and FDNY activity levels respond more to broad systemic shifts, such as changes in population movement and social activity—than to each other directly. The year 2016 appears as an outlier, likely reflecting operational or reporting differences. Overall, the lack of strong correlation aligns with the fact that OEM logs all fire-related calls, while FDNY BFI investigates only a subset of cases, resulting in diverging but somewhat parallel trends over time.
