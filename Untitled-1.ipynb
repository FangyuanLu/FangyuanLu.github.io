## About the Data

This project uses two datasets from the New York City Department of Transportation (NYC DOT) to explore how benches are distributed across the city.

### 1. City Bench Locations (Historical)

This dataset contains the full set of CityBench installations across New York City, including their coordinates, borough, community district, installation year, and whether a location is part of a Business Improvement District (BID). CityBenches are intended to support pedestrians—especially seniors and transit riders—by providing more places to sit.

**Key columns used:**
- BoroName — borough  
- CounDist — community district  
- InstallYr — installation year  
- BID — whether the bench is located in a Business Improvement District  

### 2. Bus Stop Shelters

This dataset contains the locations of NYC bus stop shelters, including their borough, bus routes served, and community district. While this dataset does not list all bus stops, bus shelters are typically placed in high-use or high-traffic locations, making them a reasonable proxy for bus activity in each neighborhood.

**Key columns used:**
- Borough — borough  
- CounDist — community district
- Shelter_ID

Both datasets include **Community District (CounDist)**, allowing me to compare the number of benches and the number of bus stop shelters in each district using simple key-based merges. This makes it possible to test whether districts with more transit stops also receive more benches.

## Initial Data Processing

```python
import pandas as pd

# Load datasets
benches = pd.read_csv('/mnt/data/City_Bench_Locations_(Historical)_20251130.csv')
bus = pd.read_csv('/mnt/data/Bus_Stop_Shelter_20251130.csv')

# --- Select only the desired columns ---
benches_sel = benches[['CounDist', 'BoroName', 'InstlYear', 'BID']].dropna(subset=['CounDist'])
bus_sel = bus[['CounDist', 'BoroName', 'Shelter_ID']].dropna(subset=['CounDist'])

# --- Group: count benches and bus shelters per Community District ---
benches_cd = benches_sel.groupby(['CounDist', 'BoroName']).size().reset_index(name='bench_count')
bus_cd = bus_sel.groupby(['CounDist', 'BoroName']).size().reset_index(name='bus_stop_count')

# --- Merge on CounDist + BoroName ---
merged = benches_cd.merge(bus_cd, on=['CounDist', 'BoroName'], how='inner')

merged.head()
```
